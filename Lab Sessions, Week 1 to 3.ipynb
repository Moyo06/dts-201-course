```python
#import statistics

import statistics as stat
```

```python
#find median of ages

print(stat.median(ages))
```

Yes, the mean is higher(26.6) than median(25.5).

Because, the mean is sensitive to extreme values, however the median represents the middle point and ignores these extream values(outliers).

When the mean and median disagree, it's telling you that your data is not perfectly symmetrical and contains outliers.

```python
#standardising state
df["State"] = df["State"].str.strip().str.title()
```

```python
#coercing marketing spend to numeric
df["Marketing Spend"] = pd.to_numeric(df["Marketing Spend"], errors="coerce")
```

```python
#simple random
srs = df.sample(n=20,  random_state=7)

#stratified, 20% of each state
strat = df.groupby("State", group_keys=False ) \
      .apply (
          lambda g: g.sample(frac=0.2))
```

```python
#comapring mean of r&d and marketing spend, with their populations and samples
compare = pd.DataFrame({
    "population": df[["Marketing Spend", "R&D Spend"]].mean(),
    "sample": srs[["Marketing Spend", "R&D Spend"]].mean()
})
print(compare.round(3))
```

R&D Spend is much closer. The sample and population's mean has a difference of 2354.758 while for Marketing Spend, the difference is about 32126.14(which is very large).

Why?
Marketing Spend has large standard deviation(123051.179029) which is nearly 3 times than R&D Spend's standard deviation. When a data has massive range, a small random samole is likely to pull extreme values that swerve the sample mean from the true populaton mean.
